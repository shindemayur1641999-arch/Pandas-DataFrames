```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python
units_skew = df["Units_Sold"].skew()
print(units_skew)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df["Units_Sold"], bins=20)
plt.title("Units_Sold distribution",fontweight="bold",color="red")
plt.xlabel("Units_Sold",fontweight="bold",color="red")
plt.figure(figsize=(2,2))
plt.ylabel("Frequency")
plt.show()
```

```python
plt.bar(df["Category"],df["Units_Sold"])
plt.xlabel("Category")
plt.ylabel("Units_sold")
plt.title('Unit Sold as per Category')
```

```python
plt.hist(df["Units_Sold"], bins=20)
plt.title("Units Sold Distribution")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")
plt.show()
```
